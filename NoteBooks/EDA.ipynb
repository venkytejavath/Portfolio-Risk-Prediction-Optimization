# Importing Libs

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
```

```python
portfolio_data=joblib.load("portfolio_data.pkl")
returns=joblib.load("Returns.pkl")
```

# Correlation

```python
corr=portfolio_data.corr(numeric_only=True)
plt.figure(figsize=(8,5))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("CORRELATION OF PORTFOLIO DATA")
plt.show()
```

The correlation heatmap shows the relationship between the portfolio asset allocations and the portfolio volatility. Most asset correlations are weak, indicating that the allocations are largely independent of one another. Among all assets, VBTLX has the strongest negative correlation with volatility (-0.86), suggesting that increasing its allocation significantly reduces portfolio risk. In contrast, VDE exhibits a strong positive correlation with volatility (0.70), indicating that higher exposure to this asset increases portfolio risk. The remaining assets (VTSAX, VTIAX, VGSLX, OPGSX, and BTC-USD) show only weak correlations with volatility, implying a relatively smaller influence on the overall portfolio risk.

# Count Plot

```python
sns.countplot(portfolio_data["Risk_category"])
plt.show()
```

The bar chart shows that the portfolio dataset is well balanced across the four risk categories: Low, Medium, Moderate, and High. Each category contains approximately 2,500 portfolio samples, indicating an equal distribution of data. This balanced class distribution is beneficial for machine learning because it prevents the model from becoming biased toward any particular risk category and helps improve the reliability and fairness of the classification results.

# HistPLlots

```python


returns.hist(
    figsize=(15, 10),
    bins=50,
    edgecolor="black"
)

plt.suptitle("Distribution of Daily Returns for Each Asset")
plt.tight_layout()
plt.show()
```

The histograms illustrate the distribution of daily returns for the seven assets in the portfolio. Most assets exhibit bell-shaped distributions centered close to zero, indicating that daily returns are generally small and fluctuate around the mean. BTC-USD shows the widest spread with the largest positive and negative returns, reflecting its high volatility compared to the other assets. In contrast, VBTLX, VTSAX, VTIAX, and VGSLX display relatively narrow distributions, indicating more stable daily returns. OPGSX and VDE exhibit moderate variability, with occasional extreme returns. Overall, the distributions highlight the differences in risk profiles among the assets, with Bitcoin being the most volatile and the bond fund (VBTLX) among the least volatile.

# BOXPlots

```python
sns.boxenplot(portfolio_data)
```

The boxplot compares the distribution of portfolio weights assigned to each asset across the randomly generated portfolios, along with the resulting portfolio volatility. VBTLX, VDE, VTIAX, and VTSAX exhibit the widest interquartile ranges and several high-value outliers, indicating greater variation in their portfolio allocations. In contrast, BTC-USD, OPGSX, and VGSLX have relatively smaller spreads, reflecting more consistent allocation ranges. The Volatility boxplot shows a moderate spread with fewer extreme values, suggesting that although asset allocations vary considerably, the resulting portfolio risk remains within a relatively stable range. Overall, the boxplot highlights the diversity of asset weight distributions while demonstrating controlled variability in portfolio volatility.
